## 2. Chargement : dataset final et taux de change, dans le cheminement

Le taux de change varie pays par pays : il rejoint donc les variables du clustering dès la préparation, comme convenu en session. Passage au logarithme, car le niveau brut d'un taux est un artefact de dénomination (185 yens et 0,85 livre ne disent rien de la santé des monnaies). Zone euro : taux 1, logarithme 0, ce qui se lit comme une absence de risque de change pour un exportateur français.


```python
df = pd.read_csv(DATA_PROCESSED / 'dataset_p11_clean.csv')

# la disponibilité de protéines était corrélée à 0,99 avec la version en kilos : c'était la même
# information à une conversion d'unité près. On l'avait retirée au notebook 1 après lecture de la
# matrice de corrélations. Si le CSV n'a pas encore été régénéré, je refais le retrait ici pour
# rester sur mes 13 variables structurelles.
cols_prot = [c for c in df.columns if 'disponibilit' in c.lower() and 'prot' in c.lower()]
if cols_prot:
    df = df.drop(columns=cols_prot)
    print('Retrait de la disponibilité de protéines (redondante avec la version en kilos)')

print(f'Dataset : {len(df)} pays, {df.shape[1]} colonnes ({df.shape[1] - 2} variables analytiques)')

# taux de change par pays, produit par le notebook 1 (section 8)
fx = pd.read_csv(DATA_PROCESSED / 'fx_par_pays.csv')
df = df.merge(fx[['iso3', 'devise', 'taux_pour_1_eur']], on='iso3', how='left')

# log du taux : atténue l'artefact de dénomination (185 yens et 0,85 livre ne se comparent pas bruts)
df['fe_change_log'] = np.log(df['taux_pour_1_eur'])

# les rares devises non cotées reçoivent la médiane (ici la table couvre tous les pays)
manquants = df['fe_change_log'].isna().sum()
df['fe_change_log'] = df['fe_change_log'].fillna(df['fe_change_log'].median())
print(f'Taux joint sur iso3 : {df["taux_pour_1_eur"].notna().sum()} pays cotés, {manquants} imputés à la médiane')
print(f'Zone euro (log = 0) : {(df["fe_change_log"] == 0).sum()} pays')
df[['pays', 'devise', 'taux_pour_1_eur', 'fe_change_log']].head()
```

## 3. Sélection des variables actives et standardisation

13 variables structurelles plus le change en log : 14 variables actives. On écarte les identifiants, la devise (texte) et le taux brut (remplacé par son log).

Normalisation retenue : log d'abord pour les variables multiplicatives (population, potentiel de marché, change), puis centrage-réduction (z-score) pour que chaque variable compte à égalité, quelle que soit son unité. C'est le prérequis de l'ACP, qui travaille sur les corrélations. Concrètement, on centre et on réduit chaque variable avec le StandardScaler : toutes sont ramenées dans des échelles de grandeur similaires, moyenne 0 et écart-type 1.


```python
# variables actives = tout sauf identifiants, devise texte et taux brut (on garde son log)
vars_actives = [c for c in df.columns if c not in ['iso3', 'pays', 'devise', 'taux_pour_1_eur']]
# attendu : 14 variables (13 structurelles + fe_change_log)
print(f'{len(vars_actives)} variables actives :')
for v in vars_actives: print(f'  - {v}')

X = df[vars_actives].copy()

# Transformation log pour les variables très asymétriques (population)
for c in ['population_2017_milliers', 'fe_potentiel_marche_musd']:
    if c in X.columns:
        X[c] = np.log1p(X[c])

# Standardisation
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled = pd.DataFrame(X_scaled, columns=vars_actives, index=df['pays'])

print(f'\nVérification standardisation : moyenne {X_scaled.mean().mean():.4f}, std {X_scaled.std().mean():.4f}')
```

## 4. Réalisation de l'ACP

```python
pca = PCA(n_components=len(vars_actives))
X_pca = pca.fit_transform(X_scaled)

inertie = pca.explained_variance_ratio_ * 100
inertie_cum = inertie.cumsum()

tab_inertie = pd.DataFrame({
    'CP': [f'CP{i+1}' for i in range(len(inertie))],
    'Valeur propre': pca.explained_variance_.round(3),
    'Inertie (%)': inertie.round(2),
    'Inertie cumulée (%)': inertie_cum.round(2),
})
tab_inertie
```

### 4.1 Choix du nombre de composantes (éboulis des valeurs propres)

```python
fig, ax = plt.subplots(figsize=(11, 6))
x_axis = np.arange(1, len(inertie) + 1)
ax.bar(x_axis, inertie, color='#2E75B6', alpha=0.7, label='Inertie individuelle')
ax.set_xlabel('Composante principale')
ax.set_ylabel('Inertie expliquée (%)', color='#2E75B6')
ax.tick_params(axis='y', labelcolor='#2E75B6')
ax.set_xticks(x_axis)

ax2 = ax.twinx()
ax2.plot(x_axis, inertie_cum, marker='o', color='#C00000', linewidth=2, label='Inertie cumulée')
ax2.axhline(80, color='gray', linestyle='--', alpha=0.5, label='Seuil 80 %')
ax2.set_ylabel('Inertie cumulée (%)', color='#C00000')
ax2.tick_params(axis='y', labelcolor='#C00000')
ax2.set_ylim(0, 105)

for i, (ind, cum) in enumerate(zip(inertie, inertie_cum)):
    ax.text(i+1, ind + 0.5, f'{ind:.1f}%', ha='center', fontsize=8.5)

plt.title('Éboulis des valeurs propres')
plt.tight_layout()
plt.savefig(FIG / '06_eboulis.png', dpi=120, bbox_inches='tight')
plt.show()

# Détermination du nombre de CP à conserver
n_cp_kaiser = sum(pca.explained_variance_ > 1)
n_cp_80 = int(np.argmax(inertie_cum >= 80) + 1)
print(f'\nCritère de Kaiser (val. propre > 1) : {n_cp_kaiser} CP')
print(f"Critère 80 % d'inertie : {n_cp_80} CP")
n_cp = max(min(n_cp_kaiser, 5), 3)
print(f'\nNombre de CP retenues pour le clustering : {n_cp}')
```

## 5. Cercle des corrélations

Le cercle projette les variables initiales sur un plan factoriel. Plus une flèche est longue (proche du cercle de rayon 1), mieux la variable est représentée. Deux flèches dans la même direction = variables fortement corrélées ; opposées = anti-corrélées ; perpendiculaires = indépendantes. Deux plans sont tracés pour couvrir les quatre composantes retenues : CP1-CP2, puis CP3-CP4.

```python
def cercle_correlations(pca, features, axes=(0, 1), figsize=(9, 9), title_suffix=''):
    fig, ax = plt.subplots(figsize=figsize)
    coords = pca.components_[list(axes)].T * np.sqrt(pca.explained_variance_[list(axes)])
    cercle = plt.Circle((0, 0), 1, color='#1F4E79', fill=False, linewidth=1.5)
    ax.add_patch(cercle)
    for i, feat in enumerate(features):
        ax.arrow(0, 0, coords[i, 0], coords[i, 1],
                 head_width=0.03, head_length=0.04, fc='#2E75B6', ec='#2E75B6', alpha=0.85)
        # Raccourcir les noms longs pour la lisibilité
        label = feat[:40] + '...' if len(feat) > 40 else feat
        ax.text(coords[i, 0]*1.12, coords[i, 1]*1.12, label,
                fontsize=7.5, ha='center', va='center', color='#1F1F1F')
    ax.axhline(0, color='gray', alpha=0.3, lw=0.8)
    ax.axvline(0, color='gray', alpha=0.3, lw=0.8)
    ax.set_xlim(-1.3, 1.3); ax.set_ylim(-1.3, 1.3); ax.set_aspect('equal')
    ax.set_xlabel(f'CP{axes[0]+1} ({pca.explained_variance_ratio_[axes[0]]*100:.1f} %)')
    ax.set_ylabel(f'CP{axes[1]+1} ({pca.explained_variance_ratio_[axes[1]]*100:.1f} %)')
    ax.set_title(f'Cercle des corrélations - Plan CP{axes[0]+1}-CP{axes[1]+1}{title_suffix}')
    plt.tight_layout()
    return fig

fig = cercle_correlations(pca, vars_actives, axes=(0, 1))
plt.savefig(FIG / '07_cercle_correlations_12.png', dpi=120, bbox_inches='tight')
plt.show()

if n_cp >= 4:
    fig = cercle_correlations(pca, vars_actives, axes=(2, 3))
    plt.savefig(FIG / '08_cercle_correlations_34.png', dpi=120, bbox_inches='tight')
    plt.show()
```

## 6. Projection des individus (pays) sur le plan factoriel

```python
def projection_individus(X_pca, pays, iso, axes=(0, 1), figsize=(13, 9)):
    fig, ax = plt.subplots(figsize=figsize)
    ax.scatter(X_pca[:, axes[0]], X_pca[:, axes[1]], s=40, alpha=0.55, color='#2E75B6')
    # Annoter les pays remarquables : extrêmes sur CP1 et CP2
    idx = np.unique(np.concatenate([
        np.argsort(X_pca[:, axes[0]])[-10:],
        np.argsort(X_pca[:, axes[0]])[:5],
        np.argsort(X_pca[:, axes[1]])[-7:],
        np.argsort(X_pca[:, axes[1]])[:5],
    ]))
    for i in idx:
        ax.annotate(iso[i], (X_pca[i, axes[0]], X_pca[i, axes[1]]),
                    fontsize=8.5, alpha=0.95, color='#1F1F1F')
    ax.axhline(0, color='gray', alpha=0.3, lw=0.8)
    ax.axvline(0, color='gray', alpha=0.3, lw=0.8)
    ax.set_xlabel(f'CP{axes[0]+1}')
    ax.set_ylabel(f'CP{axes[1]+1}')
    ax.set_title(f'Projection des pays sur le plan CP{axes[0]+1}-CP{axes[1]+1}')
    plt.tight_layout()
    return fig

fig = projection_individus(X_pca, df['pays'].values, df['iso3'].values)
plt.savefig(FIG / '09_projection_individus.png', dpi=120, bbox_inches='tight')
plt.show()
```

## 7. Contributions des variables aux axes

La contribution mesure le poids de chaque variable dans la construction d'une composante. Elle permet de baptiser la composante avec un nom métier.

```python
loadings = pca.components_.T * np.sqrt(pca.explained_variance_)
contrib_var = (loadings ** 2)
contrib_var = contrib_var / contrib_var.sum(axis=0) * 100

df_contrib = pd.DataFrame(contrib_var, index=vars_actives,
                           columns=[f'CP{i+1}' for i in range(len(vars_actives))])

print('Principaux contributeurs aux 4 premières composantes :')
for i in range(min(4, n_cp)):
    top = df_contrib[f'CP{i+1}'].sort_values(ascending=False).head(5)
    print(f'\nCP{i+1} ({pca.explained_variance_ratio_[i]*100:.1f} %) :')
    print(top.round(2))

# Heatmap
fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(df_contrib.iloc[:, :min(5, len(vars_actives))], annot=True, fmt='.1f',
            cmap='Blues', cbar_kws={'label': 'Contribution (%)'}, ax=ax)
ax.set_title('Contribution des variables aux premières CP')
plt.tight_layout()
plt.savefig(FIG / '10_contributions.png', dpi=120, bbox_inches='tight')
plt.show()
```

## 8. Sauvegarde des coordonnées factorielles pour le clustering

Le clustering se fera sur ces coordonnées plutôt que sur les variables brutes : sur les brutes, des variables corrélées comptent double dans les distances, et les axes au-delà du quatrième portent surtout du bruit. L'ACP fournit une entrée décorrélée et débruitée ; la caractérisation reviendra ensuite aux variables réelles.

```python
df_acp = pd.DataFrame(X_pca[:, :n_cp], columns=[f'CP{i+1}' for i in range(n_cp)])
df_acp.insert(0, 'iso3', df['iso3'].values)
df_acp.insert(1, 'pays', df['pays'].values)
df_acp.to_csv(DATA_PROCESSED / 'dataset_acp.csv', index=False)
print(f'Coordonnées factorielles sauvegardées : {df_acp.shape}')
df_acp.head()
```

## 9. Classification Ascendante Hiérarchique (CAH)

La CAH fusionne progressivement les paires de pays/clusters les plus proches. La méthode de Ward minimise l'augmentation de l'inertie intra-classe à chaque fusion - méthode robuste et la plus utilisée. Le dendrogramme permet de lire visuellement le bon nombre de clusters.

```python
X_clust = df_acp[[c for c in df_acp.columns if c.startswith('CP')]].values
Z = linkage(X_clust, method='ward', metric='euclidean')

# hauteur de coupe pour K = 4 : a mi-chemin entre la 4e et la 3e derniere fusion. Le seuil est
# recalcule a chaque execution, il ne peut donc pas se desynchroniser de l'arbre effectivement trace.
seuil_k4 = (Z[-4, 2] + Z[-3, 2]) / 2

fig, ax = plt.subplots(figsize=(18, 8))
dendrogram(Z, labels=df_acp['iso3'].values, leaf_rotation=90, leaf_font_size=6,
           color_threshold=seuil_k4, above_threshold_color='gray', ax=ax)
ax.set_title('Dendrogramme - méthode de Ward (CAH sur coordonnées ACP)')
ax.set_xlabel('Pays (codes ISO3)')
ax.set_ylabel("Distance d'agrégation")
ax.axhline(seuil_k4, color='red', linestyle='--', alpha=0.7,
           label=f'Coupe à K=4 (hauteur {seuil_k4:.1f})')
ax.legend()
plt.tight_layout()
plt.savefig(FIG / '11_dendrogramme.png', dpi=120, bbox_inches='tight')
plt.show()

print(f'Hauteur de coupe pour K = 4 : {seuil_k4:.2f}, '
      f'entre la 4e fusion ({Z[-4, 2]:.2f}) et la 3e ({Z[-3, 2]:.2f})')
```

### 9.1 Distances de fusion (aide au choix de K en CAH)

La distance de fusion est la distance entre clusters qu'on lit sur l'axe des y du dendrogramme : plus elle est grande, plus les groupes réunis sont différents. Tracer ces distances à part met en évidence le saut qui justifie la coupe.

```python
fig, ax = plt.subplots(figsize=(11, 5))
n_show = 25
ax.plot(range(1, n_show+1), Z[-n_show:, 2][::-1], marker='o', color='#1F4E79')
ax.set_xlabel('Nombre de clusters (lecture droite-gauche du dendrogramme)')
ax.set_ylabel("Distance d'agrégation")
ax.set_title('Distances de fusion (méthode du coude pour CAH)')
ax.grid(alpha=0.4)
plt.tight_layout()
plt.savefig(FIG / '12_coude_cah.png', dpi=120, bbox_inches='tight')
plt.show()
```

### 9.2 Coupe du dendrogramme a K=4

```python
K_cah = 4
df_acp['cluster_cah'] = fcluster(Z, t=K_cah, criterion='maxclust')
print('Distribution des pays par cluster CAH :')
print(df_acp['cluster_cah'].value_counts().sort_index())
```

## 10. K-Means : choix de K via coude et silhouette

```python
K_range = range(2, 11)
inerties = []
silhouettes = []

for k in K_range:
    km = KMeans(n_clusters=k, init='k-means++', n_init=20, random_state=42)
    labels = km.fit_predict(X_clust)
    inerties.append(km.inertia_)
    silhouettes.append(silhouette_score(X_clust, labels))

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
axes[0].plot(list(K_range), inerties, marker='o', color='#1F4E79')
axes[0].set_xlabel('K (nombre de clusters)')
axes[0].set_ylabel('Inertie intra-classe')
axes[0].set_title('Méthode du coude')
axes[0].grid(alpha=0.4)

axes[1].plot(list(K_range), silhouettes, marker='o', color='#C00000')
axes[1].set_xlabel('K')
axes[1].set_ylabel('Coefficient de silhouette')
axes[1].set_title('Coefficient de silhouette')
axes[1].grid(alpha=0.4)

plt.tight_layout()
plt.savefig(FIG / '13_choix_k.png', dpi=120, bbox_inches='tight')
plt.show()

best_k_silhouette = list(K_range)[int(np.argmax(silhouettes))]
print(f'Meilleur K selon silhouette : {best_k_silhouette}')
```

### 10.1 Application du K-Means avec le K retenu

**Justification du choix K = 4** : le coefficient de silhouette est assez plat entre K = 2 et K = 5 (il culmine légèrement à K = 3), il ne tranche donc pas vraiment ; or deux ou trois groupes resteraient trop grossiers pour isoler un segment premium exportable. Je retiens K = 4 car :
1. c'est le nombre vers lequel pointe la coupe du dendrogramme de la CAH (saut net des distances de fusion) ;
2. la silhouette à K = 4 reste dans le même mouchoir que son maximum, pas d'effondrement du score ;
3. quatre profils de marchés, c'est une granularité lisible et actionnable pour le COMEX, avec un cluster cible exploitable commercialement.

```python
K_kmeans = 4
kmeans = KMeans(n_clusters=K_kmeans, init='k-means++', n_init=50, random_state=42)
df_acp['cluster_kmeans'] = kmeans.fit_predict(X_clust) + 1
print('Distribution des pays par cluster K-Means :')
print(df_acp['cluster_kmeans'].value_counts().sort_index())
```

## 11. Visualisation des clusters sur le plan factoriel

```python
from sklearn.preprocessing import minmax_scale
fig, axes = plt.subplots(1, 2, figsize=(18, 8))
palette = sns.color_palette('Set2', max(K_cah, K_kmeans))

for ax, col, title in [(axes[0], 'cluster_cah', 'Clusters CAH'),
                        (axes[1], 'cluster_kmeans', 'Clusters K-Means')]:
    for cluster_id in sorted(df_acp[col].unique()):
        sub = df_acp[df_acp[col] == cluster_id]
        ax.scatter(sub['CP1'], sub['CP2'], s=70, alpha=0.75,
                   color=palette[cluster_id-1], label=f'Cluster {cluster_id}')
        for _, row in sub.iterrows():
            ax.annotate(row['iso3'], (row['CP1'], row['CP2']), fontsize=7, alpha=0.85)
    ax.axhline(0, color='gray', alpha=0.3, lw=0.8)
    ax.axvline(0, color='gray', alpha=0.3, lw=0.8)
    ax.set_xlabel(f'CP1 ({pca.explained_variance_ratio_[0]*100:.1f} %)')
    ax.set_ylabel(f'CP2 ({pca.explained_variance_ratio_[1]*100:.1f} %)')
    ax.set_title(title)
    ax.legend(loc='best', frameon=True)

plt.tight_layout()
plt.savefig(FIG / '14_clusters_plan_factoriel.png', dpi=120, bbox_inches='tight')
plt.show()
```

## 12. Comparaison CAH vs K-Means (matrice de confusion + ARI)

```python
tab = pd.crosstab(df_acp['cluster_cah'], df_acp['cluster_kmeans'],
                  rownames=['CAH'], colnames=['K-Means'])
print('Tableau croisé :')
print(tab)

ari = adjusted_rand_score(df_acp['cluster_cah'], df_acp['cluster_kmeans'])
nmi = normalized_mutual_info_score(df_acp['cluster_cah'], df_acp['cluster_kmeans'])
print(f'\nAdjusted Rand Index  : {ari:.3f}')
print(f'Normalized Mutual Info: {nmi:.3f}')

fig, ax = plt.subplots(figsize=(7, 6))
sns.heatmap(tab, annot=True, fmt='d', cmap='Blues', ax=ax)
ax.set_title(f'Concordance CAH vs K-Means (ARI = {ari:.3f})')
plt.tight_layout()
plt.savefig(FIG / '15_comparaison_clusters.png', dpi=120, bbox_inches='tight')
plt.show()
```

### 12.1 Contrôle : l'effet du change sur la partition

Même K-Means sur les 13 variables structurelles seules, sans le change, puis comparaison des deux partitions par l'ARI. Un ARI élevé signifie que le change affine sans bouleverser la structure.


```python
# partition sans le change : memes étapes, variables structurelles seules
vars_struct = [v for v in vars_actives if v != 'fe_change_log']
X_struct = df[vars_struct].copy()
for c in ['population_2017_milliers', 'fe_potentiel_marche_musd']:
    if c in X_struct.columns:
        X_struct[c] = np.log1p(X_struct[c])
X_struct_scaled = StandardScaler().fit_transform(X_struct)
pca_s = PCA(n_components=n_cp).fit(X_struct_scaled)
labels_struct = KMeans(n_clusters=K_kmeans, n_init=20, random_state=42).fit_predict(pca_s.transform(X_struct_scaled))

ari_change = adjusted_rand_score(labels_struct, df_acp['cluster_kmeans'])
print(f'ARI partition sans change contre partition avec change : {ari_change:.3f}')
print('lecture : proche de 1, le change affine sans bouleverser ; bas, il restructure et il faut en discuter')
```

## 13. Caractérisation des clusters (variables originales)

## 14. Recommandation finale et export

Identification du cluster cible (forte consommation de volaille + PIB élevé + stabilité politique).

La France apparaît mécaniquement dans le classement des 21 pays du cluster (rang 9) : elle présente le profil du cluster cible, un marché volaille premium installé. Étant le marché domestique de La poule qui chante, elle est **exclue des recommandations d'export**. Le top 5 opérationnel est donc : États-Unis, Chine, Brésil, Japon, Royaume-Uni. Le Brésil, premier exportateur mondial de volaille et quasi autosuffisant (taux d'import proche de 0 %), reste dans le classement au titre de la taille de son marché intérieur, mais n'est pas retenu pour le pilote : les marchés d'entrée privilégiés restent les grands importateurs proches (Allemagne, 6e, et Royaume-Uni).

```python
# Scoring du profil premium : je note chaque cluster à partir de son profil moyen (z-scores).
# La pondération redonne du poids au pouvoir d'achat et à la consommation réelle de volaille,
# pour ne pas me faire happer par les micro-États très riches mais minuscules (c'est le scoring recalibré,
# par opposition au scoring simple qui remontait Hong Kong ou les Émirats).
import re
score = pd.Series(0.0, index=profil_std.columns)
for k in profil_std.columns:
    s = 0.0
    for var in profil_std.index:
        if 'pib_par_hab' in var:
            s += profil_std.loc[var, k] * 2.5      # pouvoir d'achat, poids le plus fort
        elif 'volailles__disponibilite_alimentaire_en_quantite' in var \
             or 'volailles__disponibilité_alimentaire_en_quantité' in var:
            s += profil_std.loc[var, k] * 1.5      # marché de la volaille déjà installe
        elif 'fe_potentiel_marche_musd' in var:
            s += profil_std.loc[var, k] * 1.5      # taille du marché
        elif 'stabilite_politique' in var:
            s += profil_std.loc[var, k] * 1.0      # sécurité opérationnelle
        elif 'volailles__importations' in var:
            s += profil_std.loc[var, k] * 1.0      # appétence à l'import
    score[k] = s

print('Score du profil premium, par cluster :')
print(score.sort_values(ascending=False))
cluster_cible = int(re.sub(r'\D', '', str(score.idxmax())))
print(f'\nCluster cible retenu (scoring recalibré) : {cluster_cible}')
```

```python
# Les pays du cluster cible, classés par potentiel de marché
df_reco = (df_full[df_full['cluster_kmeans'] == cluster_cible]
           .sort_values('fe_potentiel_marche_musd', ascending=False)
           [['pays', 'iso3', 'pib_par_hab_usd', 'population_2017_milliers',
             'fe_potentiel_marche_musd', 'fe_taux_dependance_import',
             'stabilite_politique', 'inflation']])
df_reco.to_csv(OUT / 'recommendations.csv', index=False)
print(f'Les {len(df_reco)} pays du cluster cible {cluster_cible}, classés par potentiel de marché :')
df_reco
```

### 14.1 Trois contrôles de robustesse

La recommandation repose sur trois choix discutables : avoir gardé quatre composantes plutôt que
les variables brutes, avoir laissé des variables encore corrélées entre elles, et avoir pondéré le
score de sélection à la main. Je mesure chacun de ces trois choix plutôt que de le défendre par
principe.

1. redondance résiduelle : combien de paires de variables restent corrélées après la préparation ;
2. apport de l'ACP : est-ce que travailler sur les composantes change la partition par rapport aux
   variables standardisées brutes ;
3. sensibilité du scoring : est-ce que le cluster cible tient si je fais varier les cinq poids.

```python
# Contrôle 1 : redondance résiduelle entre variables actives.
# Une ACP tolère la corrélation, c'est même ce qu'elle exploite. Je vérifie seulement qu'aucune
# paire n'est quasi identique (au-delà de 0,90), ce qui reviendrait à compter deux fois la même
# information avant la standardisation.
SEUIL_CORR = 0.55
corr_abs = X_scaled.corr().abs()
paires = [(a, b, corr_abs.loc[a, b])
          for i, a in enumerate(vars_actives)
          for b in vars_actives[i + 1:]
          if corr_abs.loc[a, b] > SEUIL_CORR]
paires.sort(key=lambda t: -t[2])

print(f'Paires de variables au-dessus de {SEUIL_CORR} en valeur absolue : {len(paires)}')
for a, b, r in paires:
    print(f'  {r:.3f}  {a} / {b}')
print(f'\nCorrélation maximale résiduelle : {paires[0][2]:.3f}' if paires else '\nAucune paire au-dessus du seuil')
print('Lecture : aucune paire ne dépasse 0,90, donc aucune variable ne fait doublon.')
print("Ces corrélations sont justement ce que l'ACP va résumer sur les premiers axes.")
```

```python
# Contrôle 2 : est-ce que l'ACP change la partition ?
# Je refais le même K-Means, avec les mêmes réglages, directement sur les 14 variables
# standardisées, sans passer par les composantes. Si les deux partitions concordent, la réduction
# de dimension a bien conservé le signal ; elle aura simplement accéléré et débruité le calcul.
labels_brut = KMeans(n_clusters=K_kmeans, n_init=50, random_state=42).fit_predict(X_scaled.values)
ari_acp = adjusted_rand_score(labels_brut, df_acp['cluster_kmeans'])

print(f'ARI entre K-Means sur variables brutes et K-Means sur composantes : {ari_acp:.3f}')
print(f'Dimensions comparées : {X_scaled.shape[1]} variables brutes contre {n_cp} composantes')
print('Lecture : une concordance élevée signifie que les 4 composantes portent la même structure')
print(f"de groupes que les {X_scaled.shape[1]} variables de départ, pour {n_cp / X_scaled.shape[1]:.0%} des colonnes.")
```

```python
# Contrôle 3 : sensibilité du scoring de sélection du cluster cible.
# Les cinq poids du scoring sont un choix métier, donc un point faible assumé. Je balaie une grille
# complète : 4 valeurs pour le pouvoir d'achat, 3 valeurs pour chacun des quatre autres critères,
# soit 4 x 3^4 = 324 combinaisons, et je regarde combien désignent le même cluster.
import itertools

GRILLE_PIB = [1.5, 2.0, 2.5, 3.0]
GRILLE_AUTRES = [0.5, 1.0, 1.5]

def cluster_selectionne(w_pib, w_dispo, w_potentiel, w_stab, w_import):
    """Rejoue le scoring de la cellule précédente avec un jeu de poids donné."""
    scores = {}
    for k in profil_std.columns:
        s = 0.0
        for var in profil_std.index:
            if 'pib_par_hab' in var:
                s += profil_std.loc[var, k] * w_pib
            elif 'volailles__disponibilite_alimentaire_en_quantite' in var \
                 or 'volailles__disponibilité_alimentaire_en_quantité' in var:
                s += profil_std.loc[var, k] * w_dispo
            elif 'fe_potentiel_marche_musd' in var:
                s += profil_std.loc[var, k] * w_potentiel
            elif 'stabilite_politique' in var:
                s += profil_std.loc[var, k] * w_stab
            elif 'volailles__importations' in var:
                s += profil_std.loc[var, k] * w_import
        scores[k] = s
    return int(re.sub(r'\D', '', str(max(scores, key=scores.get))))

combinaisons = list(itertools.product(GRILLE_PIB, GRILLE_AUTRES, GRILLE_AUTRES,
                                      GRILLE_AUTRES, GRILLE_AUTRES))
choix = [cluster_selectionne(*c) for c in combinaisons]
stables = sum(1 for c in choix if c == cluster_cible)

print(f'Combinaisons de poids testées : {len(combinaisons)}')
print(f'Combinaisons qui désignent le cluster {cluster_cible} : {stables} '
      f'({stables / len(combinaisons):.1%})')
print('\nRépartition des clusters désignés selon les poids :')
for k, n in sorted(pd.Series(choix).value_counts().items()):
    print(f'  cluster {k} : {n} combinaisons')
# les combinaisons divergentes ont-elles une signature commune ?
divergentes = pd.DataFrame([c for c, k in zip(combinaisons, choix) if k != cluster_cible],
                           columns=['pib', 'dispo', 'potentiel', 'stabilite', 'import'])
print('\nSignature des combinaisons divergentes (valeurs prises par chaque poids) :')
for col in divergentes.columns:
    valeurs = [float(v) for v in sorted(divergentes[col].unique())]
    fige = ' [poids bloqué au plancher]' if len(valeurs) == 1 else ''
    print(f'  {col:<11} {valeurs}{fige}')
print(f'\nLecture : la cible ne dépend pas du réglage fin des poids. Les {len(divergentes)} '
      'combinaisons restantes')
print("partagent toutes la même signature : taille de marché ET appétence à l'import ramenées")
print("simultanément au plancher de 0,5. Autrement dit, il faut neutraliser les deux critères de")
print('volume en même temps pour déplacer la cible, ce qui contredirait la logique commerciale.')
```
